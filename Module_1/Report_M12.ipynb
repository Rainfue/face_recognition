## Датасет для детекции

Датасет включает себя изображения мужчин и женщин:

```python
# проходимся по каждой папке
for direct in os.listdir(det_data_path):
    # проверяем, находится ли по путю папка
    if os.path.isdir(dir_path := os.path.join(det_data_path, direct)):
        print('------------------------------')
        # выводим путь 
        print(dir_path)
        # выводим кол-во файлов в папке
        print(f'Кол-во файлов в папке: {len(os.listdir(dir_path))}')
print('------------------------------')
        
```

Для того, чтобы `сократить датасет`, и `сбалансировать классы`, отберу по `5000` случайных изображений из каждого класса, и удалю все остальное

```python
# проходимся по каждой папке
for direct in os.listdir(det_data_path):
    # проверяем, находится ли по путю папка
    if os.path.isdir(dir_path := os.path.join(det_data_path, direct)):
        # получаем список всех изображений
        all_files = os.listdir(dir_path)
        # получаем список случайных 5000 изображений
        selected_files = random.sample(all_files, 5000)

        for file in tqdm(all_files, desc=f'Удаление файлов: {dir_path}', unit='image'):
            # если файл не в выбранных файлах, удаляем
            if file not in selected_files:
                os.remove(os.path.join(dir_path, file))
```

Проверяем, прошло ли все успешно:

```python
# проходимся по каждой папке
for direct in os.listdir(det_data_path):
    # проверяем, находится ли по путю папка
    if os.path.isdir(dir_path := os.path.join(det_data_path, direct)):
        print('------------------------------')
        # выводим путь 
        print(dir_path)
        # выводим кол-во файлов в папке
        print(f'Кол-во файлов в папке: {len(os.listdir(dir_path))}')
print('------------------------------')
```

Для удобоства, заново пронумерую все фотографии

```python
# проходимся по каждой папке
for direct in os.listdir(det_data_path):
    # проверяем, находится ли по путю папка
    if os.path.isdir(dir_path := os.path.join(det_data_path, direct)):
        ind = 0
        # переименовываем каждый файл
        for file in tqdm(os.listdir(dir_path), f'Переименовываем файлы: {dir_path}', unit='image'):
            os.rename(
                os.path.join(dir_path, file),
                os.path.join(dir_path, f'{direct}_{ind}.jpg')
            )
            # делаем шаг
            ind += 1
```
